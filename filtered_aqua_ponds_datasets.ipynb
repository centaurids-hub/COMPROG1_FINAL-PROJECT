```python
# Step 1: Import necessary libraries

import pandas as pd

# Load dataset
data = pd.read_csv("Aquaponds Dataset.csv")

# Check columns first
print(data.columns)
```

```python
# Step 2: Convert 'Date' column to datetime format
data['Date'] = pd.to_datetime(data['Date'])
```

```python
# Step 3: Extract month from 'Date' column and create a new 'Month' column
data['Month'] = data['Date'].dt.month
```

```python
# Here, we will analyze the 'DO' (Dissolved Oxygen) column to understand its distribution and identify any potential outliers.
print(data['DO'].describe())
```

```python
# Step 4: Filter the dataset to include only rows where 'DO' is less than 6, which may indicate low oxygen levels in the water.
(data['DO'] < 6)
```

```python
# Check the distribution of 'DO' and 'TEMP' columns, and also the counts of unique values in 'station' and 'Month' columns.

print(data['DO'].describe())
print(data['TEMP'].describe())
print(data['station'].value_counts())
print(data['Month'].value_counts())
```

```python
# Step 5: Apply the filters to create a new DataFrame with the specified conditions. Which is my Unique filter logic for the dataset.
filtered_data = data[
    (data['DO'] < 6.5) &              # slightly relaxed
    (data['TEMP'] >= 25) &            # wider range
    (data['Month'] == 6) &            # keep time filter
    (data['station'] == 'station1')   # keep location
]
```

```python
# Check the shape and first few rows of the filtered dataset
print(filtered_data.shape)
filtered_data.head()
```

```python
# Save the filtered dataset to a new CSV file
filtered_data.to_csv("filtered_dataset.csv", index=False)
```

```python
# Verify that the filters have been applied correctly by checking the unique values in the relevant columns of the filtered dataset.
print((filtered_data['DO'] < 6.5).all())
print((filtered_data['station'] == 'station1').all())
print((filtered_data['Month'] == 6).all())
```

```python
# Additional Analysis: Calculate the average temperature for each month and rank the months from highest to lowest average temperature.

import pandas as pd

# Load dataset
data = pd.read_csv("Aquaponds Dataset.csv")

# Convert Date column to datetime
data['Date'] = pd.to_datetime(data['Date'])

# Extract Month
data['Month'] = data['Date'].dt.month

# Group by Month and compute average temperature
monthly_temp = data.groupby('Month')['TEMP'].mean()

# Sort from highest to lowest temperature
monthly_temp_sorted = monthly_temp.sort_values(ascending=False)

# Display results
print("Average Temperature per Month:")
print(monthly_temp)

print("\nMonths ranked by Temperature (Highest to Lowest):")
print(monthly_temp_sorted)
```
